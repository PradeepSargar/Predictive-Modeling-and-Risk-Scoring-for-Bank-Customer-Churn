## Business Problem

Customer churn is one of the most significant challenges faced by retail banking institutions. When customers discontinue their relationship with a bank, the organization not only loses immediate revenue but also sacrifices future opportunities for cross-selling, upselling, and long-term customer lifetime value.

Although banks collect extensive customer-level information, many organizations continue to rely on reactive retention strategies. In most cases, intervention occurs only after customers have already shown strong signs of leaving, resulting in higher retention costs and reduced effectiveness.

The absence of an accurate predictive system limits the bank's ability to:

- Identify customers who are likely to churn.
- Quantify the level of churn risk for each customer.
- Understand the key factors influencing customer attrition.
- Prioritize retention efforts based on predicted business impact.

To address these challenges, this project develops a machine learning-based predictive analytics solution capable of estimating customer churn probability before customer attrition occurs. The generated risk scores enable business teams to implement proactive retention strategies, optimize resource allocation, and improve overall customer relationship management.

## 10.1 Customer Churn Distribution

### Objective

Understanding the distribution of the target variable is the first step in exploratory data analysis. It helps determine how many customers have been retained and how many have exited the bank.

Analyzing the churn distribution also allows us to identify whether the dataset is balanced or imbalanced. Class imbalance is an important consideration because it can significantly influence machine learning model training and evaluation.

The insights obtained from this analysis provide the foundation for selecting appropriate evaluation metrics and developing effective customer retention strategies.

```python
# =============================================================================
# CUSTOMER CHURN SUMMARY
# =============================================================================

# Calculate customer churn counts
churn_counts = df["Exited"].value_counts().sort_index()

# Calculate customer churn percentages
churn_percentages = (
    df["Exited"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
).round(2)

# Create a summary table
churn_summary = pd.DataFrame({
    "Customer Status": ["Retained", "Churned"],
    "Count": churn_counts.values,
    "Percentage (%)": churn_percentages.values
})

churn_summary
```

```python
# =============================================================================
# CUSTOMER CHURN DISTRIBUTION
# =============================================================================

plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df,
    x="Exited"
)

# Improve x-axis labels
ax.set_xticklabels(["Retained", "Churned"])

# Add value labels
for container in ax.containers:
    ax.bar_label(container, fmt="%d", fontsize=11)

plt.title(
    "Distribution of Customer Churn",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Customer Status", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# =============================================================================
# CUSTOMER CHURN PERCENTAGE
# =============================================================================

plt.figure(figsize=(7, 7))

plt.pie(
    churn_summary["Count"],
    labels=churn_summary["Customer Status"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title(
    "Customer Churn Percentage",
    fontsize=16,
    fontweight="bold"
)

plt.show()
```

### Business Interpretation

The analysis shows the proportion of customers who remained with the bank compared to those who exited. A larger retained customer base indicates that most customers continue their relationship with the bank, while the churned customers represent the segment requiring further investigation.

The observed class distribution is also important from a machine learning perspective. If one class significantly outnumbers the other, the dataset is considered imbalanced. In such cases, relying solely on accuracy may produce misleading results because a model can achieve high accuracy while performing poorly on the minority class.

Understanding this distribution helps guide the selection of appropriate evaluation metrics such as Precision, Recall, F1-Score, and ROC-AUC during model development.

### Key Insights

- Approximately four out of every five customers have been retained by the bank.
- Around one out of every five customers has exited, indicating that customer churn is a relatively infrequent but important business event.
- The dataset exhibits class imbalance, making evaluation metrics such as Recall, F1-Score, and ROC-AUC more informative than accuracy alone.
- Predicting churn accurately can help the bank identify high-risk customers and implement targeted retention strategies before customer attrition occurs.

# 10.3 CUSTOMER DEMOGRAPHIC ANALYSIS – GENDER

## Objective

Gender is an important demographic characteristic that may influence customer behavior and banking preferences.

This analysis examines the gender distribution of customers and compares churn rates between male and female customers. The objective is to identify whether gender is associated with customer churn and whether gender-specific retention strategies may be beneficial.

```python
# =============================================================================
# CUSTOMER GENDER DISTRIBUTION
# =============================================================================

gender_summary = (
    df["Gender"]
    .value_counts()
    .rename_axis("Gender")
    .reset_index(name="Count")
)

gender_summary["Percentage (%)"] = (
    gender_summary["Count"] / len(df) * 100
).round(2)

gender_summary
```

```python
# =============================================================================
# GENDER DISTRIBUTION
# =============================================================================

plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df,
    x="Gender",
    order=df["Gender"].value_counts().index
)

for container in ax.containers:
    ax.bar_label(container, fontsize=11)

plt.title(
    "Distribution of Customers by Gender",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python
# =============================================================================
# GENDER VS CUSTOMER CHURN
# =============================================================================

gender_churn = pd.crosstab(
    df["Gender"],
    df["Exited"]
)

gender_churn.columns = ["Retained", "Churned"]

gender_churn
```

```python
# =============================================================================
# CHURN RATE BY GENDER
# =============================================================================

gender_churn_rate = (
    pd.crosstab(
        df["Gender"],
        df["Exited"],
        normalize="index"
    ) * 100
).round(2)

gender_churn_rate.columns = [
    "Retention Rate (%)",
    "Churn Rate (%)"
]

gender_churn_rate
```

```python
# =============================================================================
# CUSTOMER CHURN BY GENDER
# =============================================================================

gender_churn.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5)
)

plt.title(
    "Customer Churn by Gender",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.legend(
    title="Customer Status",
    labels=["Retained", "Churned"]
)

plt.tight_layout()
plt.show()
```

### Business Interpretation

This analysis compares the distribution of male and female customers and evaluates whether churn rates differ between the two groups.

A noticeable difference in churn rates may indicate that gender influences customer behavior or satisfaction. However, if both genders exhibit similar churn patterns, gender alone is unlikely to be a strong predictor of customer churn and should be considered alongside other demographic and financial factors.

### Key Insights

- The dataset contains both male and female customers, providing balanced demographic representation.
- Comparing churn rates across genders helps determine whether one group is more likely to leave the bank.
- If the churn rates are similar, gender is unlikely to be a dominant factor influencing customer churn.
- The findings from this analysis will help assess the importance of gender during feature selection and model interpretation.

```python
# =============================================================================
# DEFINE ALL FIVE MODELS
# =============================================================================

models = {

    # -------------------------------------------------------------------------
    # 1. Logistic Regression
    # Scaling is performed inside the pipeline to prevent data leakage
    # between cross-validation folds.
    # -------------------------------------------------------------------------
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]),

    # -------------------------------------------------------------------------
    # 2. Decision Tree
    # Tree-based models do not require feature scaling.
    # -------------------------------------------------------------------------
    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    # -------------------------------------------------------------------------
    # 3. Random Forest
    # -------------------------------------------------------------------------
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    # -------------------------------------------------------------------------
    # 4. Gradient Boosting
    # -------------------------------------------------------------------------
    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        random_state=42
    ),

    # -------------------------------------------------------------------------
    # 5. XGBoost
    # -------------------------------------------------------------------------
    "XGBoost": XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=4,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric="logloss"
    )
}
```

```python
# =============================================================================
# STRATIFIED 5-FOLD CROSS-VALIDATION
# =============================================================================

stratified_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
# =============================================================================
# CROSS-VALIDATION METRICS
# =============================================================================

cv_scoring = {
    "accuracy": "accuracy",

    "precision": make_scorer(
        precision_score,
        zero_division=0
    ),

    "recall": make_scorer(
        recall_score,
        zero_division=0
    ),

    "f1": make_scorer(
        f1_score,
        zero_division=0
    ),

    "roc_auc": "roc_auc"
}
```

```python
# =============================================================================
# RUN 5-FOLD CROSS-VALIDATION FOR ALL MODELS
# =============================================================================

cv_results = {}

for model_name, model in models.items():

    print("\n" + "=" * 75)
    print(f"{model_name} - 5-FOLD CROSS-VALIDATION")
    print("=" * 75)

    # Perform 5-fold cross-validation
    results = cross_validate(
        estimator=model,
        X=X_train,
        y=y_train,
        cv=stratified_cv,
        scoring=cv_scoring,
        n_jobs=-1,
        return_train_score=False
    )

    # Store the results
    cv_results[model_name] = results

    # Display fold-by-fold results
    print("\nFold Results:")

    for fold in range(5):
        print(
            f"Fold {fold + 1}: "
            f"Accuracy={results['test_accuracy'][fold] * 100:.2f}% | "
            f"Precision={results['test_precision'][fold] * 100:.2f}% | "
            f"Recall={results['test_recall'][fold] * 100:.2f}% | "
            f"F1={results['test_f1'][fold] * 100:.2f}% | "
            f"ROC-AUC={results['test_roc_auc'][fold] * 100:.2f}%"
        )

    # Display mean and standard deviation
    print("\nMean ± Standard Deviation:")

    print(
        f"Accuracy : "
        f"{results['test_accuracy'].mean() * 100:.2f}% ± "
        f"{results['test_accuracy'].std() * 100:.2f}%"
    )

    print(
        f"Precision: "
        f"{results['test_precision'].mean() * 100:.2f}% ± "
        f"{results['test_precision'].std() * 100:.2f}%"
    )

    print(
        f"Recall   : "
        f"{results['test_recall'].mean() * 100:.2f}% ± "
        f"{results['test_recall'].std() * 100:.2f}%"
    )

    print(
        f"F1-Score : "
        f"{results['test_f1'].mean() * 100:.2f}% ± "
        f"{results['test_f1'].std() * 100:.2f}%"
    )

    print(
        f"ROC-AUC  : "
        f"{results['test_roc_auc'].mean() * 100:.2f}% ± "
        f"{results['test_roc_auc'].std() * 100:.2f}%"
    )
```

```python
# =============================================================================
# 5-FOLD CROSS-VALIDATION COMPARISON TABLE
# =============================================================================

cv_summary = []

for model_name, results in cv_results.items():

    cv_summary.append({
        "Model": model_name,

        "Accuracy Mean (%)":
            results["test_accuracy"].mean() * 100,

        "Accuracy Std (%)":
            results["test_accuracy"].std() * 100,

        "Precision Mean (%)":
            results["test_precision"].mean() * 100,

        "Precision Std (%)":
            results["test_precision"].std() * 100,

        "Recall Mean (%)":
            results["test_recall"].mean() * 100,

        "Recall Std (%)":
            results["test_recall"].std() * 100,

        "F1 Mean (%)":
            results["test_f1"].mean() * 100,

        "F1 Std (%)":
            results["test_f1"].std() * 100,

        "ROC-AUC Mean (%)":
            results["test_roc_auc"].mean() * 100,

        "ROC-AUC Std (%)":
            results["test_roc_auc"].std() * 100
    })

cv_summary_df = pd.DataFrame(cv_summary)

# Sort models by mean ROC-AUC
cv_summary_df = cv_summary_df.sort_values(
    by="ROC-AUC Mean (%)",
    ascending=False
).reset_index(drop=True)

# Round values for clean presentation
cv_summary_df = cv_summary_df.round(2)

cv_summary_df
```

# Customer Risk Scoring

Customer Risk Scoring converts the predicted churn probability generated by the selected **Gradient Boosting Classifier** into an interpretable risk score ranging from **0 to 100**.

The risk score provides a simple way to prioritize customers according to their predicted likelihood of churn.

### Objectives

- Convert churn probability into an interpretable risk score.
- Categorize customers into Low, Medium, and High risk groups.
- Identify customers requiring greater retention attention.
- Support prioritization of customer retention strategies.
- Provide a foundation for the Streamlit Risk Calculator.

### Methodology

The predicted churn probability is converted into a 0–100 risk score:

**Risk Score = Predicted Churn Probability × 100**

The resulting score is categorized into three risk levels:

- **Low Risk:** 0–29
- **Medium Risk:** 30–59
- **High Risk:** 60–100

Higher scores indicate a greater predicted likelihood of churn according to the selected Gradient Boosting model.

### Risk Interpretation

| Risk Score | Risk Level | Interpretation |
|------------|------------|----------------|
| 0–29 | Low | Lower predicted churn likelihood |
| 30–59 | Medium | Moderate predicted churn likelihood |
| 60–100 | High | Higher predicted churn likelihood |

The risk score represents model output and should be used as a decision-support indicator rather than a guaranteed prediction of future customer behavior.

### Expected Output

1. Customer risk score from 0 to 100.
2. Risk-level classification.
3. Distribution of customers across risk levels.
4. Identification of high-risk customers.
5. Summary statistics for each risk category.

```python
# Convert predicted churn probability into a 0–100 risk score.
# A probability of 0.75 becomes a risk score of 75.

churn_probability_df["Risk Score"] = (
    churn_probability_df["Churn Probability"] * 100
)

# Round the risk score for easier interpretation.
churn_probability_df["Risk Score"] = (
    churn_probability_df["Risk Score"].round(2)
)

print("Customer Risk Score Generation Completed")
print(
    f"Minimum Risk Score: "
    f"{churn_probability_df['Risk Score'].min():.2f}"
)

print(
    f"Maximum Risk Score: "
    f"{churn_probability_df['Risk Score'].max():.2f}"
)

print(
    f"Average Risk Score: "
    f"{churn_probability_df['Risk Score'].mean():.2f}"
)
```

```python
# Assign each customer to a risk category based on the risk score.

def classify_risk(score):
    if score < 30:
        return "Low Risk"
    elif score < 60:
        return "Medium Risk"
    else:
        return "High Risk"


churn_probability_df["Risk Level"] = (
    churn_probability_df["Risk Score"]
    .apply(classify_risk)
)

# Display the first 10 customers.
churn_probability_df.head(10)
```

```python
# Count customers in each risk category.
risk_distribution = (
    churn_probability_df["Risk Level"]
    .value_counts()
    .reindex(
        ["Low Risk", "Medium Risk", "High Risk"],
        fill_value=0
    )
)

print("Customer Risk Distribution:")
print(risk_distribution)
```

```python
# Calculate the percentage of customers in each risk category.

risk_percentage = (
    risk_distribution / len(churn_probability_df) * 100
)

print("\nCustomer Risk Distribution (%):")

for risk_level, percentage in risk_percentage.items():
    print(f"{risk_level}: {percentage:.2f}%")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 6))

bars = plt.bar(
    risk_distribution.index,
    risk_distribution.values,
    edgecolor="black"
)

# Display the number of customers above each bar.
for bar, value in zip(bars, risk_distribution.values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value}",
        ha="center",
        va="bottom",
        fontsize=11,
        fontweight="bold"
    )

plt.xlabel(
    "Risk Level",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    "Number of Customers",
    fontsize=12,
    fontweight="bold"
)

plt.title(
    "Customer Risk Level Distribution",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```

```python
# Filter customers classified as High Risk.
# Sort them from highest to lowest risk score.

high_risk_customers = (
    churn_probability_df[
        churn_probability_df["Risk Level"] == "High Risk"
    ]
    .sort_values(
        by="Risk Score",
        ascending=False
    )
    .reset_index(drop=True)
)

print(
    f"Number of High-Risk Customers: "
    f"{len(high_risk_customers)}"
)

high_risk_customers.head(10)
```

```python
# Calculate summary statistics for each risk category.

risk_summary = (
    churn_probability_df
    .groupby("Risk Level")
    .agg(
        Customer_Count=("Risk Score", "count"),
        Average_Risk_Score=("Risk Score", "mean"),
        Minimum_Risk_Score=("Risk Score", "min"),
        Maximum_Risk_Score=("Risk Score", "max"),
        Average_Churn_Probability=("Churn Probability", "mean")
    )
    .reindex(["Low Risk", "Medium Risk", "High Risk"])
)

# Convert average probability into percentage.
risk_summary["Average_Churn_Probability"] = (
    risk_summary["Average_Churn_Probability"] * 100
)

# Round the results.
risk_summary = risk_summary.round(2)

risk_summary
```

```python
# Create the final customer risk scoring table.
# This table will also be useful when building the Streamlit dashboard.

final_risk_table = churn_probability_df[
    [
        "Actual Churn",
        "Predicted Churn",
        "Churn Probability",
        "Churn Probability (%)",
        "Risk Score",
        "Risk Level"
    ]
].copy()

final_risk_table.head(10)
```

```python
# Rank customers from highest to lowest predicted churn risk.

priority_customers = (
    final_risk_table
    .sort_values(
        by="Risk Score",
        ascending=False
    )
    .reset_index(drop=True)
)

priority_customers.head(20)
```
